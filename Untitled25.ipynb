```python
%pip -q install chess==1.11.2 numpy pandas matplotlib
import sys, json, platform, time, random, math, shutil
from pathlib import Path
from collections import defaultdict
from dataclasses import dataclass
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
import chess
```

```python
INITIAL_FEN = "7k/8/8/2Q5/8/8/8/4K3 w - - 0 1"


@dataclass(frozen=True)
class Config:
    initial_fen: str = INITIAL_FEN
    horizon: int = 100
    gamma: float = 0.9
    alpha: float = 0.1      # TD uniquement
    epsilon: float = 0.1    # exploration constante pour Q-learning

    def __post_init__(self):
        if self.horizon < 1:
            raise ValueError("L'horizon doit être positif.")
        if not 0 <= self.gamma <= 1:
            raise ValueError("gamma doit être dans [0, 1].")
        if not 0 < self.alpha <= 1:
            raise ValueError("alpha doit être dans ]0, 1].")
        if not 0 <= self.epsilon <= 1:
            raise ValueError("epsilon doit être dans [0, 1].")
        board = chess.Board(self.initial_fen)
        expected = sorted([chess.KING, chess.KING, chess.QUEEN])
        if (not board.is_valid() or board.turn != chess.WHITE
                or sorted(p.piece_type for p in board.piece_map().values()) != expected
                or len(board.pieces(chess.QUEEN, chess.WHITE)) != 1
                or board.castling_rights or board.ep_square is not None):
            raise ValueError("La position initiale doit être une KQK légale, aux blancs.")


def encode_state(board, remaining):
    return board.board_fen(), "white" if board.turn == chess.WHITE else "black", int(remaining)


def terminal_reason(board, truncated=False):
    # Pas de is_game_over() : les règles de répétition sont exclues.
    if board.is_checkmate():
        return "mat" if board.turn == chess.BLACK else "defaite"
    if not board.pieces(chess.QUEEN, chess.WHITE):
        return "capture_dame"
    if board.is_stalemate():
        return "pat"
    return "horizon" if truncated else None


def reward_for_reason(reason):
    return 1.0 if reason == "mat" else -1.0 if reason in ("capture_dame", "defaite") else 0.0


def legal_actions(board):
    actions = sorted(board.legal_moves, key=lambda m: m.uci())
    if not actions:
        raise ValueError("Aucune action dans un état terminal.")
    return actions


def choose_action(board, remaining, q, policy, rng, epsilon=0.1):
    """random | greedy | epsilon_greedy. Q=0 pour les couples inconnus, égalités tirées au hasard."""
    actions = legal_actions(board)
    if policy == "random":
        return rng.choice(actions)
    if policy not in ("greedy", "epsilon_greedy"):
        raise ValueError("Politique inconnue.")
    if policy == "epsilon_greedy" and rng.random() < epsilon:
        return rng.choice(actions)
    state = encode_state(board, remaining)
    values = [q.get((state, m.uci()), 0.0) for m in actions]
    best = max(values)
    return rng.choice([m for m, v in zip(actions, values) if v == best])


def play_transition(board, move, step, config, rng):
    """Coup blanc + réponse noire. Retourne (reward, raison, coups UCI joués)."""
    assert move in board.legal_moves
    board.push(move)
    played = [move.uci()]
    reason = terminal_reason(board)
    if reason is None:
        black = rng.choice(legal_actions(board))
        board.push(black)
        played.append(black.uci())
        truncated = step + 1 == config.horizon
        reason = terminal_reason(board, truncated)
    return reward_for_reason(reason), reason, played


CONFIG = Config()
SEEDS = (0, 1, 2, 3, 4)
CHECKPOINTS = (100, 1000, 5000, 10000)
N_EVAL = 1000
print(CONFIG)
display(chess.Board(CONFIG.initial_fen))
```

```python
# ------------------------------- Algorithmes TD -------------------------------
def td_target(reward, next_value, gamma, done):
    """Cible TD : r si terminal, sinon r + gamma * valeur suivante."""
    return reward if done else reward + gamma * next_value


def td_update(q, counts, key, target, alpha):
    """Q(s,a) <- Q(s,a) + alpha * (target - Q(s,a)). Retourne l'erreur TD."""
    old_value = q.get(key, 0.0)
    td_error = target - old_value
    q[key] = old_value + alpha * td_error
    counts[key] = counts.get(key, 0) + 1
    return td_error


def train_episode_td(q, counts, policy, algo, rng, config, trace=None):
    """Entraîne un épisode avec SARSA ou Q-learning. Retourne l'erreur TD absolue moyenne."""
    if algo not in ("sarsa", "q_learning"):
        raise ValueError("algo doit être 'sarsa' ou 'q_learning'.")

    board = chess.Board(config.initial_fen)
    if terminal_reason(board):
        return 0.0

    errors = []
    remaining = config.horizon
    state = encode_state(board, remaining)
    move = choose_action(board, remaining, q, policy, rng, config.epsilon)

    for step in range(config.horizon):
        reward, reason, _ = play_transition(board, move, step, config, rng)
        done = reason is not None
        next_state = next_move = None
        next_value = 0.0

        if not done:
            remaining -= 1
            next_state = encode_state(board, remaining)
            # SARSA choisit a' avant la mise à jour ; cette action sera jouée au pas suivant.
            next_move = choose_action(board, remaining, q, policy, rng, config.epsilon)
            if algo == "sarsa":
                next_value = q.get((next_state, next_move.uci()), 0.0)
            else:
                next_value = max(
                    q.get((next_state, action.uci()), 0.0)
                    for action in legal_actions(board)
                )

        key = (state, move.uci())
        target = td_target(reward, next_value, config.gamma, done)
        td_error = td_update(q, counts, key, target, config.alpha)
        errors.append(abs(td_error))

        if trace is not None:
            trace.append((state, move.uci(), reward, next_state,
                          next_move.uci() if next_move else None, done))
        if done:
            break
        state, move = next_state, next_move

    return sum(errors) / len(errors) if errors else 0.0


# ------------------------------- Génération d'épisodes pour l'évaluation -------------------------------
def generate_episode(q, policy, rng, config):
    """Épisode sans apprentissage : liste (état, action, récompense) et résultat."""
    board = chess.Board(config.initial_fen)
    episode, moves = [], []
    reason = terminal_reason(board)
    if reason:
        return episode, {"outcome": reason, "steps": 0, "plies": 0, "moves": moves}

    for step in range(config.horizon):
        remaining = config.horizon - step
        state = encode_state(board, remaining)
        move = choose_action(board, remaining, q, policy, rng, config.epsilon)
        reward, reason, played = play_transition(board, move, step, config, rng)
        moves.extend(played)
        episode.append((state, move.uci(), reward))
        if reason is not None:
            break
    return episode, {"outcome": reason, "steps": len(episode), "plies": len(moves), "moves": moves}


def discounted_return(episode, gamma):
    """Retour actualisé de l'épisode, calculé à rebours."""
    g = 0.0
    for _, _, reward in reversed(episode):
        g = reward + gamma * g
    return g


def evaluate_policy(q, policy, n_episodes, seed, config):
    """Évalue une Q-table figée ; aucune mise à jour de Q n'est effectuée."""
    if n_episodes < 1:
        raise ValueError("n_episodes doit être positif.")
    rng = random.Random(seed)
    totals = defaultdict(int)
    returns, win_steps = [], []
    for _ in range(n_episodes):
        episode, result = generate_episode(q, policy, rng, config)
        totals[result["outcome"]] += 1
        returns.append(discounted_return(episode, config.gamma))
        if result["outcome"] == "mat":
            win_steps.append(result["steps"])
    out = {f"rate_{k}": totals[k] / n_episodes
           for k in ("mat", "capture_dame", "pat", "horizon", "defaite")}
    out.update(mean_return=sum(returns) / n_episodes,
               mean_win_steps=sum(win_steps) / len(win_steps) if win_steps else math.nan,
               n_eval=n_episodes, n_wins=len(win_steps))
    return out


# ------------------------------- Expériences TD uniquement -------------------------------
# SARSA : prediction d'une politique aléatoire fixe.
# Q-learning : control, comportement epsilon-greedy et évaluation greedy.
TD_EXPERIMENTS = (
    ("td_prediction_sarsa", "sarsa", "random", "random"),
    ("td_control_q_learning", "q_learning", "epsilon_greedy", "greedy"),
)


def run_td_experiments(config, seeds, checkpoints, n_eval, progress=print):
    checkpoints = sorted(set(checkpoints))
    if not checkpoints or checkpoints[0] < 1:
        raise ValueError("Les paliers doivent être positifs.")
    metrics_rows, q_rows = [], []
    board0 = chess.Board(config.initial_fen)
    initial_state = encode_state(board0, config.horizon)
    initial_actions = [m.uci() for m in legal_actions(board0)]

    for seed in seeds:
        eval_seed = 100000 + seed
        baseline = evaluate_policy({}, "random", n_eval, eval_seed, config)
        metrics_rows.append(dict(experiment="baseline_random", seed=seed, episodes=0,
                                 q_size=0, train_seconds=0.0, mean_abs_td_error=math.nan,
                                 **baseline))
        for name, algo, train_policy, eval_policy in TD_EXPERIMENTS:
            q, counts = {}, {}
            rng = random.Random(seed)
            train_seconds, error_window = 0.0, []
            for episode_no in range(1, checkpoints[-1] + 1):
                t0 = time.perf_counter()
                error = train_episode_td(q, counts, train_policy, algo, rng, config)
                train_seconds += time.perf_counter() - t0
                error_window.append(error)

                if episode_no in checkpoints:
                    metrics = evaluate_policy(q, eval_policy, n_eval, eval_seed, config)
                    metrics_rows.append(dict(
                        experiment=name, seed=seed, episodes=episode_no,
                        q_size=len(q), train_seconds=train_seconds,
                        mean_abs_td_error=float(np.mean(error_window)), **metrics))
                    error_window = []
                    for action in initial_actions:
                        key = (initial_state, action)
                        q_rows.append(dict(experiment=name, seed=seed, episodes=episode_no,
                                           action=action, q=q.get(key, 0.0),
                                           visits=counts.get(key, 0)))
                    if progress:
                        progress(f"{name} | graine {seed} | {episode_no} épisodes | "
                                 f"mat={metrics['rate_mat']:.1%} | Q={len(q):,} | "
                                 f"entraînement={train_seconds:.1f}s")
    return metrics_rows, q_rows
```

```python
import unittest

class TestTDKQK(unittest.TestCase):
    def test_initial_position(self):
        board = chess.Board(CONFIG.initial_fen)
        self.assertTrue(board.is_valid())
        self.assertEqual(board.turn, chess.WHITE)
        self.assertEqual(len(board.piece_map()), 3)

    def test_td_update(self):
        q, counts = {("s", "a"): 0.2}, {}
        target = td_target(0.0, 0.8, 0.9, False)
        error = td_update(q, counts, ("s", "a"), target, 0.1)
        self.assertAlmostEqual(error, 0.52)
        self.assertAlmostEqual(q[("s", "a")], 0.252)

    def test_terminal_target(self):
        self.assertEqual(td_target(1.0, 999.0, 0.9, True), 1.0)
        self.assertAlmostEqual(td_target(0.0, 1.0, 0.9, False), 0.9)

    def test_sarsa_uses_next_action_played(self):
        trace = []
        train_episode_td({}, {}, "random", "sarsa", random.Random(3), Config(horizon=8), trace)
        self.assertTrue(trace)
        for current, following in zip(trace, trace[1:]):
            self.assertEqual(current[3], following[0])
            self.assertEqual(current[4], following[1])
        self.assertTrue(trace[-1][5])

    def test_q_learning_runs(self):
        q, counts = {}, {}
        train_episode_td(q, counts, "epsilon_greedy", "q_learning",
                         random.Random(1), Config(horizon=8))
        self.assertTrue(q)
        self.assertTrue(all(-1.0 <= value <= 1.0 for value in q.values()))

    def test_evaluation_does_not_update_q(self):
        q = {(encode_state(chess.Board(CONFIG.initial_fen), 5), "c5c7"): 0.5}
        before = q.copy()
        evaluate_policy(q, "greedy", 10, 1, Config(horizon=5))
        self.assertEqual(q, before)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestTDKQK)
test_result = unittest.TextTestRunner(verbosity=2).run(suite)
assert test_result.wasSuccessful(), "Corriger les tests TD avant l'entraînement."
```

```python
started = time.perf_counter()
metrics_rows, q_rows = run_td_experiments(CONFIG, SEEDS, CHECKPOINTS, N_EVAL)
elapsed_seconds = time.perf_counter() - started
metrics_df = pd.DataFrame(metrics_rows)
q_df = pd.DataFrame(q_rows)
print(f"Exécution TD terminée en {elapsed_seconds / 60:.1f} minutes")
display(metrics_df.head())
```

```python
metric_columns = ["rate_mat", "rate_capture_dame", "rate_pat", "rate_horizon",
                  "rate_defaite", "mean_return", "mean_win_steps", "q_size",
                  "train_seconds", "mean_abs_td_error"]
summary = metrics_df.groupby(["experiment", "episodes"])[metric_columns].agg(["mean", "std"])
display(summary.round(4))
q_summary = q_df.groupby(["experiment", "episodes", "action"])[["q", "visits"]].agg(["mean", "std"])
display(q_summary.round(4))

# Évolution du taux de mat et du retour actualisé
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
labels = {"baseline_random": "Référence aléatoire",
          "td_prediction_sarsa": "SARSA (prediction)",
          "td_control_q_learning": "Q-learning (control)"}
for name, label in labels.items():
    data = metrics_df[metrics_df.experiment == name].groupby("episodes")
    for ax, metric, title in [(axes[0], "rate_mat", "Taux de mat"),
                              (axes[1], "mean_return", "Retour moyen actualisé")]:
        grouped = data[metric].agg(["mean", "std"])
        ax.errorbar(grouped.index, grouped["mean"], yerr=grouped["std"].fillna(0),
                    marker="o", capsize=3, label=label)
        ax.set_xscale("log")
        ax.set_xlabel("Épisodes d'entraînement")
        ax.set_title(title)
        ax.grid(alpha=.25)
        ax.legend(fontsize=8)
fig.tight_layout()
display(fig)

# Mesures finales et Q-values des actions légales depuis la position initiale
final = metrics_df[metrics_df.episodes == max(CHECKPOINTS)]
final_summary = final.groupby("experiment")[["train_seconds", "q_size", "rate_mat",
                                                "rate_capture_dame", "mean_return"]].agg(["mean", "std"])
display(final_summary.round(4))
fig_q, ax = plt.subplots(figsize=(10, 4))
for name in ("td_prediction_sarsa", "td_control_q_learning"):
    data = q_df[q_df.experiment == name].groupby("episodes")["q"].mean()
    ax.plot(data.index, data.values, marker="o", label=labels[name])
ax.set_xscale("log")
ax.set_xlabel("Épisodes d'entraînement")
ax.set_ylabel("Q moyenne des coups initiaux")
ax.set_title("Évolution des Q-values à la position initiale")
ax.grid(alpha=.25)
ax.legend()
fig_q.tight_layout()
display(fig_q)
```
